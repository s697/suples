```python
# Derivadas parciales para regresión cúbica modificada
# y = a*x^3 + b*x^2 + d

def der_parcial_a(xs: list, ys: list) -> tuple[float, float, float, float]:
    """
    Derivada parcial respecto al parámetro a.

    Ecuación:
    c_a*a + c_b*b + c_d*d = c_ind
    """

    c_a = sum(xi**6 for xi in xs)       # x^3*x^3
    c_b = sum(xi**5 for xi in xs)       # x^3*x^2
    c_d = sum(xi**3 for xi in xs)       # x^3*1
    c_ind = sum((xi**3)*yi for xi, yi in zip(xs, ys))

    return (c_a, c_b, c_d, c_ind)



def der_parcial_b(xs: list, ys: list) -> tuple[float, float, float, float]:
    """
    Derivada parcial respecto al parámetro b.
    """

    c_a = sum(xi**5 for xi in xs)       # x^2*x^3
    c_b = sum(xi**4 for xi in xs)       # x^2*x^2
    c_d = sum(xi**2 for xi in xs)       # x^2*1
    c_ind = sum((xi**2)*yi for xi, yi in zip(xs, ys))

    return (c_a, c_b, c_d, c_ind)



def der_parcial_d(xs: list, ys: list) -> tuple[float, float, float, float]:
    """
    Derivada parcial respecto al término independiente d.
    """

    c_a = sum(xi**3 for xi in xs)       # 1*x^3
    c_b = sum(xi**2 for xi in xs)       # 1*x^2
    c_d = len(xs)                       # 1*1
    c_ind = sum(ys)

    return (c_a, c_b, c_d, c_ind)
```
