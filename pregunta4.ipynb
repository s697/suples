```python
import math

# Constante alpha para aproximación de Briggs: 1 / ln(10)
alpha = 1 / math.log(10)

def metodo_briggs(x, raices=5, decimales=6):
    """Calcula log10(x) aplicando 'raices' consecutivas redondeando a 'decimales'."""
    val = float(x)
    for _ in range(raices):
        val = round(math.sqrt(val), decimales)
    r = val - 1
    return (2**raices) * alpha * r

def error_relativo(aprox, exacto):
    return abs((exacto - aprox) / exacto)

# 1. Valores base
log1 = 0.0
log2 = metodo_briggs(2)
log3 = metodo_briggs(3)

# 2. Propiedades
log4 = 2 * log2
log5 = 1 - log2
log6_opcion1 = log2 + log3
log6_opcion2 = (7 + alpha * 0.0077696) / 9
log7 = metodo_briggs(7)
log8 = 3 * log2
log9 = 2 * log3
log10 = 1.0

# Resultados con sus errores relativos
datos = [
    (1, log1, math.log10(1)),
    (2, log2, math.log10(2)),
    (3, log3, math.log10(3)),
    (4, log4, math.log10(4)),
    (5, log5, math.log10(5)),
    ("6 (Opción 1: 2×3)", log6_opcion1, math.log10(6)),
    ("6 (Opción 2: 6⁹)", log6_opcion2, math.log10(6)),
    (7, log7, math.log10(7)),
    (8, log8, math.log10(8)),
    (9, log9, math.log10(9)),
    (10, log10, math.log10(10)),
]

print(f"{'x':<20} | {'Logaritmo':<12} | {'Error Relativo':<15}")
print("-" * 52)
for x, aprox, exacto in datos:
    err = 0.0 if exacto == 0 else error_relativo(aprox, exacto)
    print(f"{str(x):<20} | {aprox:<12.6f} | {err:<15.6e}")
```
